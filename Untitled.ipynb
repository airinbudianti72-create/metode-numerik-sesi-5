```python
import math

# Fungsi persamaan
def f(t):
    return 800 * math.exp(0.3 * t) - 2000 - 400 * t

# Interval awal
a = 5.0
b = 6.0

# Pengaturan perhitungan
toleransi = 0.0001
maks_iterasi = 100

print("=== METODE BISEKSI ===")
print("Persamaan: 800e^(0.3t) - 2000 - 400t = 0")
print()

print(f"{'Iterasi':<10}{'a':<12}{'b':<12}{'c':<12}{'f(c)':<15}")
print("-" * 61)

if f(a) * f(b) >= 0:
    print("Interval awal tidak memenuhi syarat.")
else:
    for i in range(1, maks_iterasi + 1):
        c = (a + b) / 2

        print(
            f"{i:<10}{a:<12.4f}{b:<12.4f}"
            f"{c:<12.4f}{f(c):<15.4f}"
        )

        # Menghentikan iterasi jika toleransi terpenuhi
        if abs(f(c)) < toleransi or (b - a) / 2 < toleransi:
            break

        # Menentukan interval baru
        if f(a) * f(c) < 0:
            b = c
        else:
            a = c

    print("-" * 61)
    print(f"Titik impas = {c:.4f} bulan")
```
